# Full-data model evaluation

Evaluate the checkpoints from notebook 01 on the primary and alternative data splits. Performance summaries and paired comparisons are saved under full/primary/evaluation/ and full/robust/evaluation/.
Run 00_Prepare_Data.ipynb first. Outputs are generated when this source notebook is executed.

```python
from pathlib import Path
import sys,subprocess,json,pandas as pd,platform
R=Path.cwd()
assert json.loads((R/'status.json').read_text())['phase']=='complete'
print(platform.platform())
for split in ['primary','robust']:
 q=subprocess.run([sys.executable,str(R/'full/evaluate.py'),str(R/'full'/split)],capture_output=True,text=True)
 print(split,q.stdout)
 if q.returncode: print(q.stderr)
 q.check_returncode()
```

```python
rows=[]
for split in ['primary','robust']:
 result=json.loads((R/'full'/split/'evaluation/results.json').read_text())
 for task,models in result['test'].items():
  for model,metric in models.items():rows.append(dict(split=split,task=task,model=model,MSE=metric['mse']))
display(pd.DataFrame(rows))
(R/'pipeline_complete.json').write_text(json.dumps(dict(state='complete',platform=platform.platform())))
```
